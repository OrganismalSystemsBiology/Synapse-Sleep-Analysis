```python
import json
import sys
from copy import deepcopy

def filter_small_polygons(input_path: str):
    with open(input_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    # shapesが存在しない場合の安全策
    shapes = data.get("shapes", [])
    filtered = []
    # notp=[]
    for shp in shapes:
        # shape_typeがpolygonのものを対象に、points数が1または2を抽出
        if shp.get("shape_type") == "polygon":
            pts = shp.get("points", [])
            if isinstance(pts, list) and len(pts) in (1, 2):
                filtered.append(shp)

    # 出力JSONは元構造を保ちつつ、shapesだけ差し替え
    # out = deepcopy(data)
    # out["shapes"] = filtered
    return filtered

    # with open(output_path, "w", encoding="utf-8") as f:
    #     json.dump(out, f, ensure_ascii=False, indent=2)

# if __name__ == "__main__":
    # 使い方:
    # python filter_small_polygons.py input.json output.json
    # if len(sys.argv) < 3:
    #     print("Usage: python filter_small_polygons.py <input.json> <output.json>")
    #     sys.exit(1)

    # filter_small_polygons(sys.argv[1], sys.argv[2])


from copy import deepcopy

def remove_small_polygons(input_path: str, output_path: str, min_vertices: int = 3):
    """
    LabelmeのCOCO形式JSONから、頂点数がmin_vertices未満（デフォルト3未満=1点/2点）の
    Polygonを除外し、その他はそのまま書き出します。
    """
    with open(input_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    shapes = data.get("shapes", [])
    filtered_shapes = []
    for shp in shapes:
        pts = shp.get("points", [])
        # 頂点数が3点以上のものだけ残す（1点/2点は除外）
        if isinstance(pts, list) and len(pts) >= min_vertices:
            filtered_shapes.append(shp)

    out = deepcopy(data)
    out["shapes"] = filtered_shapes

    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(out, f, ensure_ascii=False, indent=2)

```

```python
def mask_to_polygons(mask: np.ndarray, 
                     approx_eps: float = 1.0
                    ) -> list[list[tuple[int,int]]]:
    """
    2D の bool/0–1 マスクから輪郭を検出し、
    各輪郭を [(x0,y0), (x1,y1), ...] のリストとして返す。

    approx_eps: 輪郭近似の許容誤差 (ピクセル)。0 にするとそのまま生の輪郭点を返す。
    """
    # uint8 化 + 255
    mask_u8 = (mask.astype(np.uint8)) * 255

    # 外周輪郭のみを取得、点群は簡易化 (CHAIN_APPROX_SIMPLE)
    contours, _ = cv2.findContours(
        mask_u8,
        mode = cv2.RETR_EXTERNAL,
        method = cv2.CHAIN_APPROX_SIMPLE
    )

    polys: list[list[tuple[int,int]]] = []
    for cnt in contours:
        # 輪郭点を近似する (省略可)
        if approx_eps > 0:
            cnt = cv2.approxPolyDP(cnt, epsilon=approx_eps, closed=True)
        # reshape & (x,y) タプル化
        coords = cnt.reshape(-1, 2)
        poly = [(int(x), int(y)) for x, y in coords]
        # (面積が小さすぎる輪郭は破棄する場合などフィルタ可)
        if len(poly) >= 3:
            polys.append(poly)

    return polys
```

```python
import json
import numpy as np
from PIL import Image, ImageDraw
import torch
from detectron2.structures import Instances, Boxes, BitMasks

def labelme_json_to_instances(json_path):
    # 1) load LabelMe‐style JSON
    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    H = data["imageHeight"]
    W = data["imageWidth"]

    # 2) prepare lists
    boxes = []
    classes = []
    masks = []

    # 3) label → class_id map
    label2id = {"az": 0, "synapse": 1}

    # 4) iterate shapes
    for shape in data["shapes"]:
        label = shape["label"]
        pts   = shape["points"]  # List of [x, y]

        # compute bbox
        xs = [p[0] for p in pts]
        ys = [p[1] for p in pts]
        x0, y0 = min(xs), min(ys)
        x1, y1 = max(xs), max(ys)
        boxes.append([x0, y0, x1, y1])

        # class id
        classes.append(label2id[label])

        # rasterize polygon to mask
        # PIL wants (width, height) order
        img = Image.new("L", (W, H), 0)
        ImageDraw.Draw(img).polygon(pts, outline=1, fill=1)
        m = np.array(img, dtype=bool)  # H×W bool mask
        masks.append(m)

    # 5) build Instances
    inst = Instances((H, W))
    inst.pred_boxes   = Boxes(torch.tensor(boxes, dtype=torch.float32))
    inst.scores       = torch.ones(len(boxes), dtype=torch.float32)  # default score=1
    inst.pred_classes = torch.tensor(classes, dtype=torch.int64)
    # pack masks into BitMasks
    mask_tensor = torch.stack([torch.from_numpy(m.astype("uint8")) for m in masks], dim=0)
    inst.pred_masks = BitMasks(mask_tensor)

    return inst

# example usage
if __name__ == "__main__":
    json_file = "Keita #1001.json"
    instances = labelme_json_to_instances(json_file)
    print(instances)
    # Instances(num_instances=47, image_height=2626, image_width=4000, ...)
```

```python
#convert empty json file from tif image 
import os
import glob
import json
import base64
from PIL import Image

img_dir  = wsl_pyf + "/" + "train"
json_dir =  wsl_pyf + "/" + "pre_json"
os.makedirs(json_dir, exist_ok=True)

for img_path in glob.glob(os.path.join(img_dir, "*.tif")):
    # 1) 画像の base64 化
    with open(img_path, "rb") as f:
        img_b64 = base64.b64encode(f.read()).decode("utf-8")

    # 2) 画像サイズ読み込み
    with Image.open(img_path) as im:
        width, height = im.size

    # 3) JSON スケルトン作成
    data = {
        "version": "5.6.1",
        "flags": {},
        "shapes": [],        # あとで GUI で polygon を追加する
        "imagePath": os.path.basename(img_path),
        "imageData": img_b64,
        "imageHeight": height,
        "imageWidth": width
    }

    # 4) 書き出し
    base = os.path.splitext(os.path.basename(img_path))[0]
    out_file = os.path.join(json_dir, base + ".json")
    with open(out_file, "w", encoding="utf-8") as fw:
        json.dump(data, fw, ensure_ascii=False, indent=2)

    print("Generated:", out_file)
```
